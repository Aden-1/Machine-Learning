```python
import pandas as pd
import numpy as np
import csv
```

```python
data_url = "https://archive.ics.uci.edu/static/public/235/data.csv"
def load_power_data():
    csv_path = data_url
    return pd.read_csv(csv_path, sep=',', low_memory=False)
```

```python
power = load_power_data()
```
